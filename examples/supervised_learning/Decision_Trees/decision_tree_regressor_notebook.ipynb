# Decision Tree Regressor

A Decision Tree Regressor recursively splits the feature space to minimise **variance reduction**. Each leaf predicts the mean target value of training samples that reach it.

**Dataset:** California Housing — predict median house values.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
import sys
sys.path.insert(0, '.')
from decision_tree_regressor import decision_tree_regressor
np.random.seed(42)
print("Imports complete")
```

## Load & Explore the Data

```python
data = fetch_california_housing()
X, y = data.data[:500], data.target[:500]  # subset for speed
feature_names = data.feature_names

print(f"Samples:  {X.shape[0]}")
print(f"Features: {X.shape[1]} — {list(feature_names)}")
print(f"Target (median house value $100k) — min: {y.min():.2f}, max: {y.max():.2f}, mean: {y.mean():.2f}")

plt.figure(figsize=(7,4))
plt.hist(y, bins=30, color='steelblue', edgecolor='white')
plt.xlabel("Median House Value ($100k)")
plt.ylabel("Count")
plt.title("Target Distribution")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Train / Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
print(f"Train: {X_train.shape[0]} | Test: {X_test.shape[0]}")
```

## Depth vs R² Score

```python
depths = list(range(1, 11))
train_r2, test_r2 = [], []

for d in depths:
    reg = decision_tree_regressor(max_depth=d)
    reg.fit(X_train, y_train)
    train_r2.append(reg.score(X_train, y_train))
    test_r2.append(reg.score(X_test, y_test))

plt.figure(figsize=(9,5))
plt.plot(depths, train_r2, 'o-', color='steelblue', label='Train R²')
plt.plot(depths, test_r2,  's-', color='tomato',    label='Test R²')
plt.xlabel("Max Depth")
plt.ylabel("R² Score")
plt.title("Regression Tree Depth vs R²")
plt.legend()
plt.grid(True, alpha=0.4)
plt.tight_layout()
plt.show()

best_depth = depths[test_r2.index(max(test_r2))]
print(f"Best test R² {max(test_r2):.4f} at max_depth={best_depth}")
```

## Train Best Model

```python
reg = decision_tree_regressor(max_depth=5)
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)

print(f"Train R²:  {reg.score(X_train, y_train):.4f}")
print(f"Test  R²:  {reg.score(X_test,  y_test):.4f}")
print(f"Test  MSE: {reg.mean_squared_error(X_test, y_test):.4f}")
```

## Predicted vs Actual

```python
plt.figure(figsize=(7,6))
plt.scatter(y_test, y_pred, alpha=0.5, color='steelblue', edgecolors='white', linewidths=0.2, s=40)
diag = [min(y_test.min(), y_pred.min()), max(y_test.max(), y_pred.max())]
plt.plot(diag, diag, 'r--', linewidth=1.5, label='Perfect prediction')
plt.xlabel("Actual Value ($100k)")
plt.ylabel("Predicted Value ($100k)")
plt.title("Decision Tree Regressor — Predicted vs Actual")
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Residuals

```python
residuals = y_test - y_pred
plt.figure(figsize=(8,4))
plt.scatter(y_pred, residuals, alpha=0.5, color='steelblue', edgecolors='white', linewidths=0.2, s=40)
plt.axhline(0, color='tomato', linestyle='--', linewidth=1.5)
plt.xlabel("Predicted Value")
plt.ylabel("Residual")
plt.title("Residual Plot")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
print(f"Mean residual: {residuals.mean():.4f}")
print(f"Std  residual: {residuals.std():.4f}")
```

## Key Takeaways

- Regression trees make piecewise-constant predictions via recursive splits.
- **max_depth** controls complexity — deeper trees overfit training data.
- R² measures explained variance; MSE measures average squared error.
- This tree achieves meaningful predictive power without any feature scaling.
